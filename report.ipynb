This project aims to develop a Question-Answering (QA) system capable of extracting specific answers from textual context. We implemented two distinct neural architectures—a Recurrent Neural Network (LSTM) and a Transformer-based model—to evaluate their performance on an Arabic QA dataset consisting of 240 samples.

The system follows a parallel pathway architecture, designed to process the Question and the Context independently before fusing their representations to predict the start and end indices of the answer.

Pipeline Components:

Preprocessing: Normalization and tokenization were applied to the Arabic text to prepare input sequences.

Parallel Pathways: The model maintains separate processing streams for the question (intent) and context (source data) before fusion, allowing for conditioned answer extraction.

Head/Output: The final layers predict a probability distribution over the context indices to identify the start_index and end_index.

The dataset comprised 240 samples across various domains (e.g., F35 aviation, History, Biology).

Preprocessing: We mapped words to numeric indices using a custom vocabulary.

Alignment: Labels were generated by identifying the start and end word positions of the ground truth answers within the provided context.

Split: The data was partitioned into an 80/20 train/validation split to ensure robust evaluation.

We conducted comparative training over 10 epochs for both architectures.

Recurrent (LSTM) Model:

Behavior: The LSTM model demonstrated a stable training trajectory. The loss decreased consistently, and the validation loss remained aligned with the training loss, suggesting good generalization for this sequence length.

Transformer Model:

Behavior: The Transformer model converged much faster but exhibited clear signs of overfitting after the 5th epoch, where training loss continued to decline while validation loss began to rise.

nterpretation: The Transformer's high capacity allowed it to memorize the training data rapidly. Given the dataset size (240 samples), the LSTM was more robust, whereas the Transformer would likely require regularization (such as Dropout) or a larger dataset to reach its potential.

Metrics and Evaluation
We utilized Exact Match (EM) to evaluate strict accuracy and F1-Score to measure semantic overlap. These metrics are critical because simple accuracy fails to capture the complexity of span-based extraction where minor variations in particle or phrasing matter.

Limitations and Future Work
Empty Predictions: During inference, the model occasionally produces empty results when the predicted start_index exceeds the end_index. This indicates the need for a post-processing constraint to enforce end >= start.

Data Constraints: The current dataset size is relatively small for Transformer architectures, leading to the observed overfitting.

Context Length: The model currently truncates inputs to 50 tokens, which may lead to information loss for longer context passages.
